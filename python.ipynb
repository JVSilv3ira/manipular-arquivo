```python
# media de 4 numeros
def media(a, b, c, d):
    return (a + b + c + d) / 4

valores = input("Digite 4 números separados por espaço: ")
numeros = list(map(float, valores.split()))
print("A média é:", media(*numeros))
```

```python
# achar pares entre 10 numeros
def achar_pares(numeros):
    pares = [num for num in numeros if num % 2 == 0 and num <= 10]
    return pares
    for num in numeros:
        if num % 2 == 0 and num <= 10:
            pares.append(num)
    return pares
valores = input("Digite números separados por espaço: ")
numeros = list(map(int, valores.split()))
pares = achar_pares(numeros)
print("Números pares:", pares)
```

```python
# receber numeros até digitar 0, dps somar eles e contar quantos foram digitados
def contar_numeros():
    conta = 0
    total = 0
    while True:
        num = int(input("Digite um numero (digite 0 para parar): "))
        if num == 0:
            break
        conta += 1
        total += num
    return conta, total

conta, total = contar_numeros()
print(f"Quantidade de numeros digitados: {conta}")
print(f"Soma total dos numeros: {total}")
```
